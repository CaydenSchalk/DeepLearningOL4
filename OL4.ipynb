```python
import numpy as np
import matplotlib.pyplot as plt
```

## Translation

```python
lstm, simple = np.loadtxt('translation-results.txt', unpack=True)
data = [simple, lstm]

bp = plt.boxplot(data,notch=True)
plt.title('Translation: Validation Accuracy')
plt.ylabel('Accuracy (P)')
plt.xticks([1,2],['Simple','LSTM'])
plt.ylim([0.625,0.725])
plt.show()
```

## Parity

```python
lstm, simple, custom = np.loadtxt('parity-results.txt', unpack=True)
data = [simple, lstm, custom]

bp = plt.boxplot(data,notch=True)
plt.title('Parity: Validation Accuracy')
plt.ylabel('Accuracy (P)')
plt.xticks([1,2,3],['Simple','LSTM','Custom'])
plt.show()
```

A. LSTM on average does slightly better in the translation problem, but that doesn't hold true for the parity problem without the custom change.

B. LSTM is slightly stronger in the translation problem, but does worse than RNN in the parity problem. Training gets stuck very early for the LSTM, while the RNN doesn't really get stuck at all. In order for the LSTM to match parameters with the RNN, I had to reduce the number of layers, which may have had a play in that. LSTM overall is better at carrying information from farther places over.

C. They can represent parity, but they can't reliably learn parity. Overall it's more of an approximation, because if it found the exact solution, it would be near or at 100% accuracy by the end of training.

D. I expected it to do better than before, but not as good as it actually did. It did confirm my hypothesis, as overall when LSTM is given an auxiliary task for the encoder, it allows the encoder to not get stuck and have a better chance at learning the problem.

E. More runs will always help get a better idea of the variance overall, but 10 runs seems to be fairly stable.

F. Nothing I can think of.

G. I'm interested to learn more about why exactly LSTMs can compete with GRUs. I presume even though GRUs are newer and easier to train, they do have less expression that LSTMs can have.
